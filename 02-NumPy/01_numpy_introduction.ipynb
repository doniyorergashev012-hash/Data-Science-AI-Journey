10-mashq. List va Array farqi

Quyidagi koddan foydalan:

my_list = [1, 2, 3, 4, 5]

my_array = np.array([1, 2, 3, 4, 5])



my_list * 2 natijasini chiqar.

my_array * 2 natijasini chiqar.

Nima uchun natijalar farq qilishini izohla.

```python
my_list = [1, 2, 3, 4, 5]
my_array = np.array([1, 2, 3, 4, 5])
for son in my_list:
  print(son*2,end=" ")
```

```python
print(*my_array*2)
```

11-mashq. Vektorlashtirish

numbers = list(range(1, 101))

List yordamida har bir sonni 3 ga ko‘paytir.

Array yordamida har bir sonni 3 ga ko‘paytir.

Ikkala natijani solishtir.

```python
numbers = list(range(1, 101))
array_numbers=np.array(numbers)
for son in numbers:
  print(son*2,end=" ")
print(array_numbers*2)
```

12-mashq. Katta array bilan ishlash

my_list = list(range(100000))



Ushbu list'dan NumPy array yarat.

List elementlarini 2 ga ko‘paytir.

Array elementlarini 2 ga ko‘paytir.

Ikkala usulni %time orqali o‘lchab ko‘r.

```python
my_list = list(range(100000))
my_array=np.array(my_list)
%time my_array*2
```

```python
%time for son in my_list:son*2
```

13-mashq. Array xususiyatlarini tekshirish

numbers = [10, 20, 30, 40, 50, 60, 70, 80]



Array yarat.

.ndim, .shape, .size qiymatlarini chiqar.

Har bir xususiyat nimani anglatishini o‘z so‘zlaring bilan yoz.

```python
numbers = [10, 20, 30, 40, 50, 60, 70, 80]
num_array=np.array(numbers)
print(num_array.ndim) #Nechi o'lchamli massiv ekanini anqilash uchun .ndim funksiyasi ishlatiladi
print(num_array.size) #Nechta element bor ekanini aniqlash uchun .size funksiyasidan foydalaniladi
print(num_array.shape) #Massiv nechiga nechi ekanini yana qatorlar va ustunlar sonini korsatadi
```

15-mashq. Mini-loyiha — NumPy ma'lumotlar generatori

Quyidagi vazifalarni bitta dasturda bajar:

1. 20 ta tasodifiy butun son yarat.

2. 10 ta tasodifiy kasr son yarat.

3. 5 ta elementli np.zeros() array yarat.

4. 5 ta elementli np.ones() array yarat.

5. Har bir array'ni ekranga chiqar.

6. Har birining .size, .ndim, .shape qiymatlarini chiqar.

7. Har bir array qanday yaratilganini izohlab ber.

```python
tasodifiy_butn=np.random.randint(1,40,20)
tasodifiy_kasr=np.random.uniform(1,50,10)
nollar=np.zeros(5)
birlar=np.ones(5)
print(tasodifiy_butn.size)
print(tasodifiy_kasr.size)
print(nollar.size)
print(birlar.size)
```

```python
print(tasodifiy_butn.shape)
print(tasodifiy_kasr.shape)
print(nollar.shape)
print(birlar.shape)
```

```python
print(tasodifiy_butn.ndim)
print(tasodifiy_kasr.ndim)
print(nollar.ndim)
print(birlar.ndim)
```
